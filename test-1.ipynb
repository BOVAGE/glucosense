```python
def time_str_to_minutes(col_name, prefix):
    """Converts 'hr-1:30' → 90.0 minutes. Used to sort lag columns by recency."""
    lag_str = col_name.replace(prefix + '-', '')  # e.g. '1:30'
    h, m = lag_str.split(':')
    return int(h) * 60 + int(m)
```

```python
hr_cols = [col for col in df_work_1.columns if col.startswith('hr-')]
steps_cols = [col for col in df_work_1.columns if col.startswith('steps-')]
hr_cols_sorted = sorted(hr_cols, key=lambda col: time_str_to_minutes(col, 'hr'))
steps_cols_sorted = sorted(steps_cols, key=lambda col: time_str_to_minutes(col, 'steps'))
print("hr_cols_sorted:", hr_cols_sorted)
print("steps_cols_sorted:", steps_cols_sorted)
selected_hr_cols = hr_cols_sorted[:LOOPBACK_PERIOD_IN_MINS // 5]
selected_steps_cols = steps_cols_sorted[:LOOPBACK_PERIOD_IN_MINS // 5]
print(f"{len(selected_hr_cols)} selected_hr_cols:", selected_hr_cols)
print(f"{len(selected_steps_cols)} selected_steps_cols:", selected_steps_cols)
```
